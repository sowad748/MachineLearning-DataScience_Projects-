```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Visualization settings
plt.style.use('seaborn-v0_8')
sns.set_palette('Set2')
```

```python
df = pd.read_csv('/content/netflix_titles.csv')

df.head()
```

```python
df.info()
df.shape
```

```python
df.isnull().sum()
```

```python

# Drop rows
df.dropna()
```

```python
df['type'].value_counts()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='type')
plt.title('Movies vs TV Shows on Netflix')
plt.show()
```

```python
df['date_added'].head()
```

```python
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
df['date_added'].head()
```

```python
df.info()
```

```python
df['year_added'] = df['date_added'].dt.year
```

```python
df[['date_added', 'year_added']].head()
```

```python
content_per_year = df['year_added'].value_counts().sort_index()
content_per_year
```

```python
plt.figure(figsize=(10,5))
plt.plot(content_per_year.index, content_per_year.values, marker='o')

plt.title('Netflix Content Added Over Years')
plt.xlabel('Year')
plt.ylabel('Number of Titles Added')
plt.grid(True)

plt.show()
```

```python
type_year = df.groupby(['year_added', 'type']).size().unstack()
type_year
```

```python
type_year.plot(figsize=(10,5))

plt.title('Movies vs TV Shows Added Over Years')
plt.xlabel('Year')
plt.ylabel('Number of Titles')
plt.legend(title='Content Type')

plt.show()
```

```python
import plotly.express as px

interactive_df = (
    df.groupby(['year_added', 'type'])
    .size()
    .reset_index(name='count')
)

fig = px.line(
    interactive_df,
    x='year_added',
    y='count',
    color='type',
    title='Netflix Content Growth Over Time'
)

fig.show()
```
